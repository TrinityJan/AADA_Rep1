```python
def ass_1(lines):
    n, d, c = lines[0].split()
    # print(n, d, c)
    lines.pop(0)
    df = pd.DataFrame([[float(j) for j in i.split()] for i in lines[:-int(c)]])
    cd = [[float(j) for j in i.split()] for i in lines[-int(c):]]
    if d == '2':
        plot = df.plot(kind='scatter', x=0, y=1, c=df.columns.size-1, colormap='tab20', colorbar=False)
        # x, y = zip(*cd)
        x = [row[0] for row in cd]
        y = [row[1] for row in cd]
        plot.scatter(x, y, marker='X', s=150, c=range(int(c)), cmap='tab20', linewidths=1, edgecolors='black')
#     if d == '3':
#         df.columns = ['x', 'y', 'z', 'c']
#         sns.pairplot(df, vars=['x', 'y', 'z'], hue='c')
#
# def ass_2(lines):
#     n, d, c = lines[0].split()
#     print(n, d, c)
#     lines.pop(0)
#     df = pd.DataFrame([[float(j) for j in i.split()] for i in lines])
    # if d == '2':
    #     plot = df.plot(kind='scatter', x=0, y=1, c=df.columns.size-1, colormap='tab20', colorbar=False)
    # if d == '3':
    #     print()
    #     fig = plt.figure()
    #     ax = fig.add_subplot(111, projection='3d')
    #     xs = df.iloc[:,0].tolist()
    #     ys = df.iloc[:,1].tolist()
    #     zs = df.iloc[:,2].tolist()
    #     c = df.iloc[:,3].tolist()
    #     ax.scatter(xs, ys, zs, c=c, cmap='tab20')

plotters = {
    '1': ass_1,
    # '2': ass_2,
    # '3': ass_2
}
```

```python
def plot_file(fl):
    with open(fl) as f:
        lines = f.read().splitlines()
    a = lines[0].split()[0]
    # print(a)
    lines.pop(0)
    plotters[a](lines)
```

```python
# EXTRA: visualize the input

def ass_1_input(lines):
    n, d, c = lines[0].split()
    print(n, d, c)
    lines.pop(0)
    # Input: just n points with d coordinates, no cluster labels
    df = pd.DataFrame([[float(j) for j in i.split()] for i in lines[:int(n)]])
    if d == '2':
        df.plot(kind='scatter', x=0, y=1)
        plt.title('Input data (unlabelled)')
        plt.show()
    if d == '3':
        df.columns = ['x', 'y', 'z']
        sns.pairplot(df, vars=['x', 'y', 'z'])
        plt.suptitle('Input data (unlabelled)', y=1.02)
        plt.show()
```

```python
def plot_input_file(fl):
    with open(fl) as f:
        lines = f.read().splitlines()
    ass_1_input(lines)
```
